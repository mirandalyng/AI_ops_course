# Structured output from text

```python

from model import changeModel

from pydantic_ai import Agent 
from pydantic import BaseModel, Field
```

```python
class Person(BaseModel): 
    name: str = Field("Person first and last name")
    organization: str = Field("Company or organisation")
    subject: str = Field("The subject the person is responseble for")

class Resilience(BaseModel): 
    titel: str = Field("Pick suiteble headline from the input. Don't hallucinate")
    description: str = Field("3-5 sentences summary of the contetnt")
    contributors: list[Person]
    contact: str = Field("contact person for this event")


```

```python
with open(
    "raw_text.txt",
    encoding="utf-8"
) as file:
    raw_text = file.read()

print(raw_text)
```

```python
agent = Agent(
    model=changeModel("nvidia/nemotron-3.5-super-vl-120b-a12b"), 
    instructions="You are an expert in summarizing swedish content in swedish."
    " Take the procited content and summarize it "
    )


result = await agent.run(raw_text, output_type=Resilience)

```

```python
print(result.output)
```

```python
content = result.output

content.contributors
```

```python
content.contact
```

## Structure a program 


```python
with open(
    "raw_program.txt",
    encoding="utf-8"
) as file:
    raw_program = file.read()

print(raw_program)

```

```python

class Courses(BaseModel):
    course: str = Field("Name of the course")
    points: int = Field("Number of Yh-points")
    year: int = Field("Number of which year the course is icluded in")
    content: str = Field("What is included in the course")
    
class Program(BaseModel): 
    name: str = Field("Title of the program")
    description: str = Field("Description 4-5 sentences about the program")
    courses: list[Courses] 


result = await agent.run(raw_program, output_type=Program)
```

```python
print(result.output)
```

```python
prog = result.output
```

```python
prog.courses
```

```python
for prog in prog.courses: 
    print(prog)
```
