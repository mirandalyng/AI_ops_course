# PydanticAI with Nvidia Build

### Create a model 
- this is from the models in Nvidia Build from  
- Model name is:
    nvidia/nemotron-3.5-lightning-30b-a3b
- use either environ[] or getenv()

```python
from pydantic_ai import Agent 
# Pga Nvida inte finns som modell 
# Betyder ej att det är en öppen AI 
from pydantic_ai.models.openai import OpenAIChatModel 
from pydantic_ai.providers.openai import OpenAIProvider 
import os 
from dotenv import load_dotenv
load_dotenv()


#create model 
model = OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-lightning-30b-a3b", #this is found in nvida - choose model and copy model
    provider=OpenAIProvider(
        base_url="https://integrate.api.nvidia.com/v1",
        api_key=os.environ["NVIDIA_API_KEY"]  # from .env, getenv(api_key) funkar också 
    ),

)

model
```

```python
agent = Agent(model)
result = await agent.run("Give me a funny python joke", timeout=60)
result 
```

```python
result = await agent.run("ge mig ett skämt med c#", timeout=30)
print(result.output)
```

## Change model 


```python
# 
def changeModel (model_name): 
    OpenAIChatModel(
        # this is found in nvida - choose model and copy model
        model_name=model_name,
        provider=OpenAIProvider(
            base_url="https://integrate.api.nvidia.com/v1",
            # from .env, getenv(api_key) funkar också
            api_key=os.environ["NVIDIA_API_KEY"]
        ),

    )

model = changeModel("nvidia/nemotron-3.5-super-vl-120b-a12b")
```

```python
result = await agent.run("Berätta ett skämt om en norsk", model=model)
result
```
