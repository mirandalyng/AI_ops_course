# Llm intro 

```python
from pydantic_ai import Agent
# Pga Nvida inte finns som modell
# Betyder ej att det är en öppen AI
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv
load_dotenv()
```

```python

def changeModel(model_name):

    return OpenAIChatModel(
        # this is found in nvida - choose model and copy model
        model_name=model_name,
        provider=OpenAIProvider(
            base_url="https://integrate.api.nvidia.com/v1",
            # from .env, getenv(api_key) funkar också
            api_key=os.environ["NVIDIA_API_KEY"]
        ),

    )
```

```python
model = changeModel("nvidia/nemotron-3.5-lightning-30b-a3b")

model
```

```python
agent = Agent(model=model, system_prompt="You are a funny joking bot, answer shortly in a nerdy style, also add emojis")

prompt = "Explain what PydanticAI is and the differance betewen it and Pydantic"

result = await agent.run(prompt)

result
```

# ChefBot

```python
chef_agent = Agent(
    model = model, 
    instructions=(
        "You are a professional chef that is very good at making asian food and"
        " explaining in a pedagogical way how to create yummy dishes."
        """
                Answer in the following format:

                Description 1-2 sentences of what the dish is
    
                List of ingredients in bullet points

                Short step by step description on how to make the dish
                """ )
)


result = await chef_agent.run("Give me 3 asian soups in a list")
print(result.output)
```

```python
with open("soups.md", "w") as file: 
    file.write(result.output)
```

# Thinking 

```python
from pydantic_ai import ModelSettings 

storyteller = Agent(
    model, 
    instructions = "You are a children storyteller, answer max 3 sentences", 
    model_settings=ModelSettings(thinking="minimal", temperature=0)
)


promt = "A rabbit called Bella likes to program Python ... "

story1 =  await storyteller.run(promt)

print(story1.output)
```

# Multimode

```python
# change model 

model = changeModel("nvidia/nemotron-3.5-super-vl-120b-a12b")
```

```python
from pathlib import Path

from pydantic_ai.messages import BinaryContent

image_joker = Agent(
    model=model, 
    instructions="You are a person that can analyze an image, explain the contents and make a joke about it, max 3 sentences."
)

image_data = Path("bella.png").read_bytes()
promt = "Tell me about this dude"


result = await image_joker.run(
    user_prompt=[promt, BinaryContent(data = image_data, media_type="image/png")]
)

print(result.output) 
```
