# FireCrawl 

```python
import requests

import os 
from dotenv import load_dotenv
load_dotenv()

FIRECRAWL_API = os.environ["FIRECRAWL_API_KEY"]

url = "https://api.firecrawl.dev/v2/search"

payload = {
    "query": "what are the lastest news about Jensen Huang?", 
    "sources": [
        "web"
    ],
    "categories": [],
    "limit": 3,
    "scrapeOptions": {
        "onlyMainContent": True,
        "maxAge": 172800000,
        "parsers": [
            "pdf"
        ],
        "formats": []
    }
}

headers = {
    "Authorization": f"Bearer {FIRECRAWL_API}",
    "Content-Type": "application/json"
}

response = requests.post(url, json=payload, headers=headers)

print(response.json())

```

## Crawling

```python
from firecrawl import Firecrawl
from firecrawl.v2.types import ScrapeOptions

app = Firecrawl()

scrape_opts = ScrapeOptions(
    only_main_content=True,
    max_age=172800000,
    parsers=["pdf"],
    formats=["markdown"]
)

crawl_result = app.crawl(
    "https://www.myh.se/validering-och-seqf",
    sitemap="include",
    crawl_entire_domain=False,
    limit=3,
    scrape_options=scrape_opts
)

print(crawl_result)
```

```python
crawl_result.__dict__.keys()
```

```python
crawl_result.credits_used
```

```python
crawl_result.total
```

```python
crawl_result.data
```

```python
from pathlib import Path 
Path("myh").mkdir(exist_ok=True)
for doc in crawl_result.data: 
    metadata = doc.metadata
    with open (f"myh/{metadata.title}.md", "w") as f:
        f.write(doc.markdown)
    
```
